## Introduction

In the rapidly evolving landscape of digital assets, maintaining consistent profitability remains one of the most fundamental challenges in quantitative finance. The extreme volatility, nonlinear dependence structures, and frequent regime transitions characteristic of cryptocurrency markets often render conventional trading frameworks ineffective beyond narrow time horizons. Despite substantial progress in algorithmic trading and machine learning, many existing approaches continue to face limitations related to nonstationarity, parameter instability, and liquidity fragmentation—factors that erode robustness and predictive reliability.

Unlike traditional financial markets, cryptocurrency ecosystems operate in a less efficient and more reflexive informational environment, where feedback loops between trader behavior, leverage dynamics, and exchange microstructure frequently give rise to transient but statistically recurring inefficiencies. These inefficiencies are further amplified by the presence of retail-driven volume, uneven funding-rate arbitrage mechanisms, and periodic liquidity shocks across exchanges. As a result, market equilibrium conditions are rarely achieved, allowing a properly tuned algorithmic framework to extract consistent value from these repeating imbalances while maintaining a high degree of capital efficiency.

The digital asset domain also exhibits a structurally elevated risk profile. This arises not only from extreme volatility and fragmented liquidity but also from the pervasive use of high leverage, which magnifies both opportunity and exposure. Leverage compounds small microstructural inefficiencies into significant price dislocations, while simultaneously increasing the potential for systemic cascades during liquidation events. The confluence of these elements produces a market that is simultaneously rich in opportunity and fraught with risk—demanding sophisticated, risk-aware, and dynamically adaptive strategies to remain sustainable over time..

## Literature Review

The evolution of systematic trading research has been shaped by foundational theories in market efficiency, stochastic modeling, and behavioral dynamics. Early work by Fama (1970) and Samuelson (1965) established the efficient market hypothesis and the random walk characterization of price behavior, defining the theoretical boundaries for forecastability in asset returns. However, subsequent research in behavioral and nonlinear finance (Shiller, 1981; Lo, 2004) has demonstrated that bounded rationality, heterogeneous expectations, and structural asymmetries can lead to persistent deviations from efficiency.

Parallel developments in high-frequency and algorithmic trading have emphasized the importance of latency, execution costs, and microstructural frictions (Aldridge, 2013; Cartea, Jaimungal, and Penalva, 2015). Empirical analyses of tick-level data (Cont, 2001; Zhang, Russell, and Tsay, 2008) reveal that intraday returns often exhibit volatility clustering, mean reversion, and transient momentum bursts, providing exploitable but short-lived opportunities for systematic trading. These insights underpin much of the design philosophy behind dynamic, risk-sensitive execution frameworks

The efficient-market hypothesis (Fama, 1970) posits that all available information is immediately reflected in prices, precluding consistent excess returns. Yet subsequent work by Shiller (1981), Lo (2004), and others introduced behavioral and evolutionary perspectives that acknowledge periods of structural inefficiency and regime dependence.

High-frequency and algorithmic-trading research (Cartea et al., 2015; Aldridge, 2013) emphasizes that latency, microstructural friction, and liquidity constraints create transient inefficiencies exploitable by well-calibrated systems. Cont (2001) formalized key empirical properties of asset returns—volatility clustering, heavy tails, and autocorrelation—which are magnified in crypto markets.

Studies specific to cryptocurrencies (Dyhrberg, 2016; Baur et al., 2018; Corbet et al., 2020) describe persistent deviations from normality and inefficient cross-exchange pricing. However, most existing frameworks either rely on static econometric models or deep neural networks with limited interpretability. The DCAF model bridges these extremes by combining transparent stochastic control logic with machine-learning-based adaptive calibration..

In the context of cryptocurrencies, research remains relatively recent but expanding rapidly. Studies such as Dyhrberg (2016) and Baur, Dimpfl, and Kuck (2018) document distinctive volatility patterns and asymmetric correlations between digital and traditional assets. Further investigations (Phillip et al., 2019; Corbet et al., 2020) highlight how decentralized market microstructures, liquidity fragmentation, and exchange-specific funding mechanisms generate unique risk-return dynamics that differ markedly from those of conventional asset classes.

Most existing crypto-trading methodologies rely on static parameterizations or opaque machine learning architectures that prioritize predictive accuracy over interpretability and stability. This study departs from both extremes by proposing a parameter-rich yet analytically transparent framework rooted in stochastic process theory and adaptive control. Signal layers interact deterministically through structured rules that adjust exposure in response to volatility, drawdown, and liquidity metrics. This approach combines the clarity of rule-based systems with the adaptability of data-driven optimization, allowing stable performance across distinct volatility regimes.

By bridging these perspectives, the present work contributes to the emerging literature on adaptive trading under nonstationary market dynamics. It demonstrates that systematic and interpretable strategies can sustain superior risk-adjusted performance in an asset class traditionally regarded as unpredictable 

## Strategy architecture and design

The firs curcial component of this strategy is non traditional parameter selection and configurations.
Extensive testing and empirical validation demonstrated that a compact, single-layer parameter structure was insufficient to sustain a meaningful statistical edge, particularly under high-leverage and rapidly changing market conditions. To address this, we used a hierarchical parameter architecture and an additional configuration immposed on the parameters, consisting of multiplle layers, where each independent layer processes the outputs of the previous one. This design enables progressive refinement of both signal precision and risk modulation.

The first layer captures coarse directional features and baseline volatility characteristics; subsequent layers adaptively recalibrate thresholds, position sizing, and temporal weighting based on the preceding layer’s performance metrics. This hierarchical approach allows for dynamic scaling of returns while simultaneously stabilizing exposure and enhancing drawdown resilience.

While this depth of parameterization introduces the theoretical risk of overfitting, extensive out-of-sample validation and stress testing across multiple market regimes, liquidity conditions, and time horizons confirmed that the observed performance improvements were not artifacts of curve fitting. Instead, the results consistently reflected a genuine and repeatable statistical edge—robust to noise, leverage variation, and regime shifts.

Furthermore, cross-validation between disjoint time intervals and exchanges demonstrated that the layered model preserved rank consistency among parameter configurations, indicating structural robustness rather than incidental correlation.

However, identifying a favorable parameter combination within such a large search space—consisting of 15–20 parameters, each with multiple possible value, proved to be a formidable and time-consuming challenge. The high sensitivity of certain parameters to small perturbations, combined with the presence of multiple sequential layers, made exhaustive exploration computationally prohibitive. To accelerate the tuning process, machine learning models, particularly Bayesian optimization, were employed at various stages to efficiently navigate the parameter space and converge toward optimal configurations.

In essence, the first stage of the framework focuses on identifying a viable parameter set and determining the shortest time interval—referred to as the base period, over which the statistical advantage of that configuration remains stable and significant.

The next stage is expanding the base period to target invertment/trading horizon by adding a second analytical layer built around a Dynamic Capital Allocation Framework. This layer uses the base period performance as input and applies a martingale style optimization approach under stochastic process principles to determine optimal scaling of capital exposure through time. The framework continuously rebalances allocation weights in response to realized volatility, drawdown history, and local risk conditions, thereby targeting optimal long term growth rates. In practice, this produces annualized returns substantially higher than the simple annualization implied by the square root of n scaling rule, as the compounding dynamics and probabilistic reinvestment logic magnify geometric growth while maintaining risk discipline. The mathematical foundation of this framework draws upon stochastic control, ergodic theory, and martingale fairness conditions to ensure pathwise growth rate maximization under volatility clustering and serial correlation.

Furthermore, to prevent parameter drift and local over-adaptation, the framework incorporates a reinforcement-style learning mechanism that updates allocation parameters according to realized performance feedback. Allocations that historically improve geometric growth are reinforced through incremental weight adjustments, whereas those that degrade growth are penalized. This process allows the system to approximate the long-run solution of a stochastic dynamic programming problem without explicit re-optimization at every step.

From an economic standpoint, this architecture operationalizes the principle that capital should flow toward opportunity only when the marginal contribution to expected growth exceeds its marginal contribution to risk. Over many trading periods, this policy leads the system toward an invariant distribution of exposures that maximizes long-run compounding while maintaining resilience under stress.


## APPENDIX: The Pyton Implementation (Stage 1)

This section demostrates a Phyton implementation of our stage 1 process. The first stage of our analysis is  designed to capture the structural inefficiencies underlying short-term cryptocurrency price dynamics. The script constitutes the operational core of our framework, executing thousands of randomized simulations over minutely data to test a wide parameter space of leverage, position duration, and signal configurations. The structure of the program is intentionally modular: all proprietary mathematical logic, signal-generation algorithms, and parameter calibration routines are encapsulated within independent model files, ensuring that the public-facing simulation engine reveals no sensitive intellectual property. The primary module acts as an orchestration layer, calling the relevant proprietary functions from external model components, loading cleaned market data, and evaluating the statistical distribution of performance metrics across randomized subsamples.

The objective of this first-stage implementation was to explore and delineate the parameter space capable of producing exceptional returns over relatively short trading horizons while maintaining volatility within tolerable bounds. In this sense, the simulation sought not to optimize for risk-adjusted stability but to identify the regions in which the underlying framework exhibits its strongest raw edge. The baseline model represents what might be termed the “idealized live period,” a hypothetical environment in which the strategy operates continuously under clean execution conditions, absent unmodeled frictions or liquidity distortions. This period serves as the reference benchmark against which all subsequent refinements and friction-adjusted versions are compared.

Empirically, the early results of this stage show very high mean returns and exceptionally strong Sortino ratios, coupled with moderate Sharpe ratios that remain in a conservative range. This pattern indicates that the system consistently generates asymmetric upside performance—large positive outliers with relatively limited downside exposure. Such outcomes are characteristic of strategies that capitalize on persistent microstructural patterns but have not yet undergone full volatility management through position-sizing or dynamic capital-allocation control.

The intentionally optimistic metrics also reflect the partial omission of market frictions. Certain forms of slippage, spread expansion, and latency are inherently stochastic and cannot be captured deterministically. Yet, it would be incorrect to conclude that all frictions act strictly as performance detractors. In practice, specific frictions can enhance realized returns under favorable structural conditions. For instance, when the strategy’s entry timing systematically aligns with directional order-flow momentum, the resulting positive execution slippage can improve average fill prices. Likewise, favorable funding asymmetry—common in cryptocurrency perpetual markets—may yield a persistent tailwind when the strategy’s bias coincides with the profitable funding side. Even micro-latency effects can occasionally benefit performance if rapid local price movements occur in the trade’s favor between signal generation and execution. Thus, while uncertainty about frictions introduces variability, their aggregate impact can sometimes improve live outcomes.

In summary, the Stage 1 implementation serves as the foundation of the broader research framework. It isolates the raw statistical edge of the proprietary signal model, maps the regions of maximum profitability within the parameter landscape, and establishes a quantitative baseline from which to evolve. Later stage in the framework will progressively introduce mechanisms for dynamic capital allocation and martingale-type optimization through time, systems that will both stabilize volatility without losing edge provided by high returns  by adaptively reallocating exposure based on rolling performance and market state feedback. Together, these refinements transform the baseline prototype from a high-potential exploratory model into a robust, production-ready trading architecture.

The script below demostrates a single trial conducted with 75 random samples of each covering one month period, folowed by summary statistics directly produced by running the script. So these results are reproducble given the private models and datasets.

```python
import numpy as np
import pandas as pd

from ST_config import (
    margin, N, K, K2, K3, a, b, a2, b2, a3, b3, lev, tp, sl, d, r, h, z, L, pm, pos, SC
)
from ST_signals import add_st_features
# np.random.seed(42) # For reproducibilty

# -----------------------------
# Public Parameters (kept as in original)
# -----------------------------
start_bal  = 100_000 # Note that what really matters here is the relative size of positions 
# to the initial balance since the returns are calculated in percent terms rather than absolute dollars.
slice_sz   = N  # Here we specifically adjusted N to cover roguhly one month period.
num_slics = pm*pos #  Here each value computes a separate corresponds to a sparate onfiguration of the same slice. But for demonstration we used the worst case escenario, 1, a single 
## to reduce volatility.
M          = 75
fee        = 0.0004

# -----------------------------
# Data
# -----------------------------
raw = pd.read_csv("Coins_cleaned.csv")
df  = raw.iloc[::r].reset_index(drop=True)  # For creating different price frequencies

tickers = SC if SC else ["BTC", "ETH", "BNB", "XRP", "ADA", "LTC"]
all_stats = {}


# Simulation Engine
class Sim:
    def __init__(self, px: pd.Series):
        self.px = pd.to_numeric(px, errors="coerce").dropna().reset_index(drop=True)
        self.n  = len(self.px)

    def _close(self, p, t, px_np):
        entry    = p["entry_price"]
        notional = p["notional"]
        mrg      = p["margin"]
        side     = p["side"]
        price    = px_np[t]
        exit_fee = fee * notional

        r_unlev = (price / entry) - 1.0 if side == 1 else (entry / price) - 1.0
        r_lev   = lev * r_unlev
        pnl     = mrg * r_lev

        if r_lev <= -1.0:
            return True, 0.0, exit_fee, "LIQ"
        if r_lev >= tp:
            return True, mrg + pnl, exit_fee, "TP"
        if r_lev <= sl:
            return True, max(mrg + pnl, 0.0), exit_fee, "SL"
        if t >= p["expiry_idx"]:
            return True, max(mrg + pnl, 0.0), exit_fee, "EXP"
        return False, 0.0, 0.0, None

    def run(self):
        sr = []  
        cfgs = [(pos * m, Lval) for Lval in L for m in pm]

        for _ in range(M):
            sim_rets = []
            sim_posn = []

            for pos_sz, Lval in cfgs:
                st = np.random.randint(0, self.n - slice_sz)
                cut = self.px.iloc[st:st + slice_sz].reset_index(drop=True).to_frame("close")

                # Signals
                cut = add_st_features(
                    cut, K=K, K2=K2, K3=K3, L_val=Lval,
                    a=a, b=b, a2=a2, b2=b2, a3=a3, b3=b3, h=h
                )

                px_np   = cut["close"].to_numpy()
                n_slice = len(px_np)

                bal = float(start_bal)
                book = []
                t_cool = 0
                opened = 0

                for t in range(n_slice):
                    # Exits
                    keep = []
                    for p in book:
                        done, back, fee_out, _reason = self._close(p, t, px_np)
                        if done:
                            if back < p["margin"]:
                                t_cool = t + d
                            bal += back
                            bal -= fee_out
                        else:
                            keep.append(p)
                    book = keep

                    if t < t_cool:
                        continue

                    sig = int(cut.at[t, "signal"])
                    if sig == 0 or t + Lval >= n_slice:
                        continue

                    # Entry
                    tgt_mrg   = pos_sz
                    notion    = tgt_mrg * lev
                    fee_in    = fee * notion
                    need_cash = tgt_mrg + fee_in
                    if bal < need_cash:
                        continue

                    bal -= need_cash
                    entry_px = px_np[t]
                    book.append({
                        "entry_idx": t,
                        "expiry_idx": t + Lval,
                        "side": 1 if sig == 1 else -1,
                        "entry_price": entry_px,
                        "margin": tgt_mrg,
                        "notional": notion,
                    })
                    opened += 1

                if book:
                    t_last = n_slice - 1
                    for p in book:
                        done, back, fee_out, _ = self._close(p, t_last, px_np)
                        if not done:
                            entry  = p["entry_price"]
                            price  = px_np[t_last]
                            side   = p["side"]
                            mrg    = p["margin"]
                            notion = p["notional"]
                            r_unlev = (price / entry) - 1.0 if side == 1 else (entry / price) - 1.0
                            r_lev   = lev * r_unlev
                            pnl     = mrg * r_lev
                            back    = max(mrg + pnl, 0.0)
                            fee_out = fee * notion
                        bal += back
                        bal -= fee_out
                    book.clear()

                ret_pct = ((bal - start_bal) / start_bal) * 100.0
                sim_rets.append(ret_pct)
                sim_posn.append(opened)

            avg_ret = float(np.mean(sim_rets)) if sim_rets else 0.0
            avg_pos = float(np.mean(sim_posn)) if sim_posn else 0.0
            sr.append((avg_ret, avg_pos))

        rts = pd.Series([x[0] for x in sr])
        pns = pd.Series([x[1] for x in sr])

        mu = rts.mean()
        sd = rts.std()
        dn = rts[rts < 0].std()

        sh = mu / sd if sd and sd > 0 else float("nan")
        so = mu / dn if dn and dn > 0 else float("inf")

        pct = {f"{p}%": float(np.percentile(rts, p)) for p in range(10, 100, 10)}

        out = {
            "count": int(rts.shape[0]),
            "mean": round(float(mu), 2),
            "std":  round(float(sd), 2),
            "min":  round(float(rts.min()), 2),
            **{k: round(v, 2) for k, v in pct.items()},
            "max":  round(float(rts.max()), 2),
            "sharpe":  round(float(sh), 2) if np.isfinite(sh) else float("nan"),
            "sortino": round(float(so), 2) if np.isfinite(so) else float("inf"),
            "avg_positions": round(float(pns.mean()), 2) if len(pns) else 0.0,
        }
        return out


n_coins = len(tickers)
all_stats = {}

for i, _t in enumerate(tickers):
    print(f"\n=== Running simulations for {_t} ===")  # show coin name
    s = Sim(df[_t]).run()
    key = _t if n_coins == 1 else f"{_t}_{i+1}"  # use actual coin name as key
    all_stats[key] = s

#---------------------------------
# Report
#---------------------------------

cols = [
    "count", "mean", "std", "min",
    "10%", "20%", "30%", "40%", "50%", "60%", "70%", "80%", "90%",
    "max", "sharpe", "sortino",
]
tbl = pd.DataFrame(all_stats).T.reindex(columns=cols)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")
print("\nFinal Summary Table (Monthly):")
print(tbl)

```

## Expedted Market Frictions that are not modeled

Although the backtest offers substantial evidence of the strategy’s edge, live results are expected to trail simulated performance. This difference mainly arises from execution frictions that are difficult to model precisely in historical simulations, including:

- Order-execution latency: Even small delays between signal generation and execution can lead to less favorable fills, especially at high frequencies.

- Slippage and spread expansion: Real-time order-book fluctuations often widen spreads or cause partial fills, slightly degrading realized returns.

- Liquidity fragmentation: Cross-exchange price dispersion introduces additional transaction costs not reflected in aggregated datasets.

- Funding-rate asymmetries and exchange fees: These recurrent but variable costs accumulate and can materially impact net performance.

Despite these inevitable frictions, the ranking and directional consistency of configurations between simulated and live runs remains strong. Empirical testing over multiple assets and time segments confirms that parameter sets yielding superior backtest results continue to outperform in production, validating the statistical integrity and transferability of the model.


These findings provide strong evidence that cryptocurrency derivatives markets is highly likely to be partially inefficient due to endogenous feedback between leverage, liquidity, and trader behavior. The persistence of adaptive profitability challenges the strong form of the Efficient Market Hypothesis and supports the Adaptive Market Hypothesis (Lo, 2004). Structural asymmetries such as exchange-specific funding mechanisms, fragmented liquidity, and heterogeneous leverage access limit arbitrage efficiency and sustain exploitable return patterns.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from ST_config import (
    margin, N, K, K2, K3, a, b, a2, b2, a3, b3,
    lev, tp, sl, d, r, h, z, L, pm, pos, SC
)
from ST_signals import add_st_features

np.random.seed(42)

# -----------------------------
# Public Parameters
# -----------------------------
start_bal = 100_000
slice_sz = N
num_slics = pm * pos
M = 75
fee = 0.0004

# Approximate market friction parameters
slippage_mean = 0.0006       # ~6 bps typical spread widening
slippage_std = 0.0003
latency_prob = 0.10          # 10% of trades hit delayed fills
latency_impact = 0.001       # 0.1% average adverse fill
shock_prob = 0.02            # rare severe frictions
shock_impact = 0.003         # 0.3% impact when shock occurs

# -----------------------------
# Data
# -----------------------------
raw = pd.read_csv("Coins_cleaned.csv")
df = raw.iloc[::r].reset_index(drop=True)
tickers = SC if SC else ["BTC", "ETH", "BNB", "XRP", "ADA", "LTC"]
all_stats, returns_dict = {}, {}

# -----------------------------
# Simulation Engine
# -----------------------------
class Sim:
    def __init__(self, px: pd.Series):
        self.px = pd.to_numeric(px, errors="coerce").dropna().reset_index(drop=True)
        self.n = len(self.px)

    def _apply_market_friction(self, price):
        """Apply random execution noise to simulate real-world frictions."""
        slip = np.random.normal(slippage_mean, slippage_std)
        if np.random.rand() < latency_prob:
            slip += latency_impact
        if np.random.rand() < shock_prob:
            slip += shock_impact
        # Randomly positive or negative
        direction = 1 if np.random.rand() > 0.5 else -1
        return price * (1 + direction * slip)

    def _close(self, p, t, px_np):
        entry = p["entry_price"]
        notional = p["notional"]
        mrg = p["margin"]
        side = p["side"]

        # Apply market friction on exit as well
        price = self._apply_market_friction(px_np[t])
        exit_fee = fee * notional

        r_unlev = (price / entry) - 1.0 if side == 1 else (entry / price) - 1.0
        r_lev = lev * r_unlev
        pnl = mrg * r_lev

        if r_lev <= -1.0:
            return True, 0.0, exit_fee, "LIQ"
        if r_lev >= tp:
            return True, mrg + pnl, exit_fee, "TP"
        if r_lev <= sl:
            return True, max(mrg + pnl, 0.0), exit_fee, "SL"
        if t >= p["expiry_idx"]:
            return True, max(mrg + pnl, 0.0), exit_fee, "EXP"
        return False, 0.0, 0.0, None

    def run(self):
        sr = []
        cfgs = [(pos * m, Lval) for Lval in L for m in pm]
        all_returns = []

        for _ in range(M):
            sim_rets, sim_posn = [], []

            for pos_sz, Lval in cfgs:
                st = np.random.randint(0, self.n - slice_sz)
                cut = self.px.iloc[st:st + slice_sz].reset_index(drop=True).to_frame("close")
                cut = add_st_features(cut, K=K, K2=K2, K3=K3, L_val=Lval,
                                      a=a, b=b, a2=a2, b2=b2, a3=a3, b3=b3, h=h)
                px_np = cut["close"].to_numpy()
                n_slice = len(px_np)
                bal = float(start_bal)
                book = []
                t_cool, opened = 0, 0
                curve = []

                for t in range(n_slice):
                    # Exits
                    keep = []
                    for p in book:
                        done, back, fee_out, _ = self._close(p, t, px_np)
                        if done:
                            if back < p["margin"]:
                                t_cool = t + d
                            bal += back - fee_out
                        else:
                            keep.append(p)
                    book = keep

                    if t < t_cool:
                        curve.append(bal)
                        continue

                    sig = int(cut.at[t, "signal"])
                    if sig == 0 or t + Lval >= n_slice:
                        curve.append(bal)
                        continue

                    # Entry
                    entry_px = self._apply_market_friction(px_np[t])
                    tgt_mrg = pos_sz
                    notion = tgt_mrg * lev
                    fee_in = fee * notion
                    need_cash = tgt_mrg + fee_in
                    if bal < need_cash:
                        curve.append(bal)
                        continue

                    bal -= need_cash
                    book.append({
                        "entry_idx": t,
                        "expiry_idx": t + Lval,
                        "side": 1 if sig == 1 else -1,
                        "entry_price": entry_px,
                        "margin": tgt_mrg,
                        "notional": notion,
                    })
                    opened += 1
                    curve.append(bal)

                if book:
                    t_last = n_slice - 1
                    for p in book:
                        done, back, fee_out, _ = self._close(p, t_last, px_np)
                        if not done:
                            price = self._apply_market_friction(px_np[t_last])
                            entry = p["entry_price"]
                            side = p["side"]
                            mrg = p["margin"]
                            notion = p["notional"]
                            r_unlev = (price / entry) - 1.0 if side == 1 else (entry / price) - 1.0
                            pnl = mrg * lev * r_unlev
                            back = max(mrg + pnl, 0.0)
                            fee_out = fee * notion
                        bal += back - fee_out
                    book.clear()

                ret_pct = ((bal - start_bal) / start_bal) * 100.0
                sim_rets.append(ret_pct)
                sim_posn.append(opened)
                all_returns.append(pd.Series(np.array(curve) / start_bal - 1))

            avg_ret = np.mean(sim_rets) if sim_rets else 0.0
            avg_pos = np.mean(sim_posn) if sim_posn else 0.0
            sr.append((avg_ret, avg_pos))

        rts = pd.Series([x[0] for x in sr])
        mu, sd = rts.mean(), rts.std()
        dn = rts[rts < 0].std()
        sh = mu / sd if sd > 0 else np.nan
        so = mu / dn if dn > 0 else np.inf
        pct = {f"{p}%": float(np.percentile(rts, p)) for p in range(10, 100, 10)}

        out = {
            "count": int(rts.shape[0]),
            "mean": round(mu, 2),
            "std": round(sd, 2),
            "min": round(float(rts.min()), 2),
            **{k: round(v, 2) for k, v in pct.items()},
            "max": round(float(rts.max()), 2),
            "sharpe": round(sh, 2) if np.isfinite(sh) else np.nan,
            "sortino": round(so, 2) if np.isfinite(so) else np.inf,
            "avg_positions": round(float(np.mean([x[1] for x in sr])), 2),
            "returns": pd.concat(all_returns, axis=1).mean(axis=1) if all_returns else None,
        }
        return out


# ---------------------------------
# Run simulations and plot
# ---------------------------------
n_coins = len(tickers)
all_stats, returns_dict = {}, {}

for i, _t in enumerate(tickers):
    print(f"\n=== Running simulations for {_t} ===")
    s = Sim(df[_t]).run()
    key = _t if n_coins == 1 else f"{_t}_{i+1}"
    all_stats[key] = {k: v for k, v in s.items() if k != "returns"}
    returns_dict[key] = s.get("returns")

# ---------------------------------
# Report
# ---------------------------------
cols = ["count", "mean", "std", "min",
        "10%", "20%", "30%", "40%", "50%", "60%", "70%", "80%", "90%",
        "max", "sharpe", "sortino"]
tbl = pd.DataFrame(all_stats).T.reindex(columns=cols)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")
print("\nFinal Summary Table (Monthly):")
print(tbl)

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from ST_config import (
    margin, N, K, K2, K3, a, b, a2, b2, a3, b3,
    lev, tp, sl, d, r, h, z, L, pm, pos, SC
)
from ST_signals import add_st_features

K2 = 10
a2 =0.01
b2=0.01

# =============================
# Balanced realism parameters
# =============================
start_bal = 100_000
slice_sz  = N
M         = 75
fee       = 0.0004

# Execution and costs; calibrated to plausible live crypto conditions
slippage_mean = 0.0010      # 10 bps average slippage
slippage_std  = 0.0015      # 15 bps std
latency_prob  = 0.25        # 25% of fills get latency hit
latency_impact = 0.0008     # +8 bps when delayed
shock_prob    = 0.03        # 3% microstructure stress bars
shock_impact  = 0.0040      # +40 bps extra on shocks
volatility_scaling = True    # slippage rises with realized vol
fee_multiplier = 1.15        # modest taker bias overall
false_signal_prob = 0.05     # small adversarial noise on signals
extra_liq_penalty_bps = 10   # 10 bps of notional at liquidation; bankruptcy friction

# Simple funding model; small mean; modest variability per bar while position open
# This is a per-bar rate applied to notional; sign can be +/-; model as zero-mean noise
funding_mu   = 0.0
funding_std  = 0.00002       # 2 bps per bar at the extreme tail is very rare; mean near zero

# Optional guardrails
max_open_positions = 3       # cap concurrent positions per symbol
min_bar_vol = 0.0            # optionally skip entries when vol below threshold


# =============================
# Data
# =============================
raw = pd.read_csv("Coins_cleaned.csv")
df  = raw.iloc[::r].reset_index(drop=True)
tickers = SC if SC else ["BTC", "ETH", "BNB", "XRP", "ADA", "LTC"]


# =============================
# Side-aware execution model
# =============================
def exec_price(mid_price: float, local_vol: float, side: int, is_entry: bool) -> float:
    """
    Adverse; side-aware execution away from mid.
    side: +1 long; −1 short.
    """
    slip = np.random.normal(slippage_mean, slippage_std)
    if np.random.rand() < latency_prob:
        slip += latency_impact
    if np.random.rand() < shock_prob:
        slip += shock_impact * np.random.uniform(0.5, 1.5)
    if volatility_scaling:
        slip *= (1 + 3.0 * min(abs(local_vol), 0.04))  # gentler scaling

    slip = abs(slip)
    # Long pays up at entry; sells down at exit; Short sells down at entry; buys up at exit
    if side == 1:
        adverse_sign = +1 if is_entry else -1
    else:
        adverse_sign = -1 if is_entry else +1
    return mid_price * (1 + adverse_sign * slip)


# =============================
# Simulation Engine
# =============================
class Sim:
    def __init__(self, px: pd.Series):
        self.px = pd.to_numeric(px, errors="coerce").dropna().reset_index(drop=True)
        self.n  = len(self.px)

    def _close(self, p, t, px_np, local_vol):
        """
        Close/mark; return (done; cashflow; reason; trade_pnl)
        cashflow is amount added to balance now; includes exit fee and funding accrued.
        trade_pnl is pure PnL net of fees and funding for distribution stats; 0 when not closing.
        """
        entry = p["entry_price"]
        side  = p["side"]
        mrg   = p["margin"]
        notion = p["notional"]

        price = exec_price(px_np[t], local_vol, side=side, is_entry=False)
        exit_fee = fee_multiplier * fee * notion

        # price move
        r_unlev = (price / entry) - 1.0 if side == 1 else (entry / price) - 1.0
        r_lev   = lev * r_unlev
        pnl     = mrg * r_lev

        # funding accrued while open
        funding_cost = p.get("funding_accum", 0.0)

        # liquidation; margin gone; pay fees and liquidation penalty
        if r_lev <= -1.0:
            extra_pen = (extra_liq_penalty_bps / 10000.0) * notion
            cashflow = -(exit_fee + extra_pen + funding_cost)
            trade_pnl = cashflow  # entire loss realized here
            return True, cashflow, "LIQ", trade_pnl

        bal_change = mrg + pnl - exit_fee - funding_cost  # no flooring
        # realize and return margin plus pnl less costs
        if r_lev >= tp:
            return True, bal_change, "TP", (pnl - exit_fee - funding_cost)
        if r_lev <= sl:
            return True, bal_change, "SL", (pnl - exit_fee - funding_cost)
        if t >= p["expiry_idx"]:
            return True, bal_change, "EXP", (pnl - exit_fee - funding_cost)
        return False, 0.0, None, 0.0

    def run(self):
        cfgs = [(pos * m, Lval) for Lval in L for m in pm]

        run_stats = []
        all_returns = []
        trade_pnls = []

        for _ in range(M):
            sim_rets, sim_posn = [], []
            sim_mdds = []

            for pos_sz, Lval in cfgs:
                st = np.random.randint(0, self.n - slice_sz)
                cut = self.px.iloc[st:st + slice_sz].reset_index(drop=True).to_frame("close")

                # Signals and features
                cut = add_st_features(
                    cut, K=K, K2=K2, K3=K3, L_val=Lval,
                    a=a, b=b, a2=a2, b2=b2, a3=a3, b3=b3, h=h
                )

                # trade next bar to avoid look-ahead
                if "signal" not in cut.columns:
                    raise ValueError("add_st_features must create a 'signal' column.")
                cut["signal_trade"] = cut["signal"].shift(1).fillna(0)

                px_np   = cut["close"].to_numpy()
                n_slice = len(px_np)

                bal = float(start_bal)
                book = []
                curve = []

                t_cool = 0
                opened = 0

                for t in range(1, n_slice):
                    local_vol = abs(np.log(px_np[t] / px_np[t - 1]))

                    # accrue funding on open positions for this bar
                    if book:
                        fr = np.random.normal(funding_mu, funding_std)
                        for p in book:
                            # funding accumulates on notional; sign can be +/-; subtract from cash on exit
                            p["funding_accum"] = p.get("funding_accum", 0.0) + (fr * p["notional"])

                    # exits
                    keep = []
                    for p in book:
                        done, cashflow, reason, trade_pnl = self._close(p, t, px_np, local_vol)
                        if done:
                            if reason in ("SL", "LIQ") and cashflow <= 0.0:
                                t_cool = t + d
                            bal += cashflow
                            if trade_pnl != 0.0:
                                trade_pnls.append(trade_pnl)
                        else:
                            keep.append(p)
                    book = keep

                    if t < t_cool:
                        curve.append(bal)
                        continue

                    # entry
                    sig = int(cut.at[t, "signal_trade"])
                    if np.random.rand() < false_signal_prob and sig != 0:
                        sig *= -1
                    if sig == 0 or t + Lval >= n_slice:
                        curve.append(bal)
                        continue
                    if len(book) >= max_open_positions:
                        curve.append(bal)
                        continue
                    if local_vol < min_bar_vol:
                        curve.append(bal)
                        continue

                    side = 1 if sig == 1 else -1
                    entry_px = exec_price(px_np[t], local_vol, side=side, is_entry=True)

                    tgt_mrg = pos_sz
                    notion  = tgt_mrg * lev
                    fee_in  = fee_multiplier * fee * notion
                    need_cash = tgt_mrg + fee_in

                    if bal < need_cash:
                        curve.append(bal)
                        continue

                    bal -= need_cash
                    book.append({
                        "entry_idx": t,
                        "expiry_idx": t + Lval,
                        "side": side,
                        "entry_price": entry_px,
                        "margin": tgt_mrg,
                        "notional": notion,
                        "funding_accum": 0.0,
                    })
                    opened += 1
                    curve.append(bal)

                # force close remaining at end
                if book:
                    t_last = n_slice - 1
                    for p in book:
                        # accrue one last funding bar for symmetry
                        fr = np.random.normal(funding_mu, funding_std)
                        p["funding_accum"] = p.get("funding_accum", 0.0) + (fr * p["notional"])

                        done, cashflow, reason, trade_pnl = self._close(p, t_last, px_np, 0.01)
                        if not done:
                            # compute explicit forced exit with adverse execution
                            price = exec_price(px_np[t_last], 0.01, side=p["side"], is_entry=False)
                            mrg   = p["margin"]
                            notion = p["notional"]
                            exit_fee = fee_multiplier * fee * notion
                            r_unlev = (price / p["entry_price"]) - 1.0 if p["side"] == 1 else (p["entry_price"] / price) - 1.0
                            pnl = mrg * lev * r_unlev
                            funding_cost = p.get("funding_accum", 0.0)
                            cashflow = mrg + pnl - exit_fee - funding_cost
                            trade_pnl = (pnl - exit_fee - funding_cost)
                        bal += cashflow
                        if trade_pnl != 0.0:
                            trade_pnls.append(trade_pnl)
                    book.clear()

                # stats for this config slice
                curve = np.array(curve) if len(curve) else np.array([bal])
                ret_pct = ((bal - start_bal) / start_bal) * 100.0
                sim_rets.append(ret_pct)
                sim_posn.append(opened)

                # max drawdown on this slice
                eq = pd.Series(curve)
                roll_max = eq.cummax()
                dd = (eq - roll_max)
                mdd = dd.min()  # negative
                sim_mdds.append(float(mdd))

                all_returns.append(pd.Series(curve / start_bal - 1.0))

            # aggregate within this outer run
            rts = pd.Series(sim_rets)
            mu, sd = rts.mean(), rts.std()
            dn = rts[rts < 0].std()
            sh = mu / sd if sd > 0 else np.nan
            so = mu / dn if dn > 0 else np.nan

            run_stats.append({
                "mean": mu,
                "std": sd,
                "sharpe": sh,
                "sortino": so,
                "mdd": np.mean(sim_mdds)  # average of slice MDDs; still conservative
            })

        # final aggregation over runs
        df_rs = pd.DataFrame(run_stats)
        out = {
            "count": len(df_rs),
            "mean": round(df_rs["mean"].mean(), 2),
            "std": round(df_rs["std"].mean(), 2),
            "min": round(df_rs["mean"].min(), 2),
            "max": round(df_rs["mean"].max(), 2),
            "sharpe": round(df_rs["sharpe"].mean(), 2),
            "sortino": round(df_rs["sortino"].mean(), 2),
            "avg_mdd_abs": round(abs(df_rs["mdd"].mean()), 2),  # in currency units of balance trajectory
            "returns": pd.concat(all_returns, axis=1).mean(axis=1),
            "trade_pnls": trade_pnls,  # list for optional histogram
        }
        return out


# =============================
# Run and report
# =============================
all_stats = {}
returns_dict = {}

for _t in tickers:
    print(f"\n=== Running simulations for {_t} ===")
    s = Sim(df[_t]).run()
    # make a shallow copy without large arrays for table
    slim = {k: v for k, v in s.items() if k not in ("returns", "trade_pnls")}
    all_stats[_t] = slim
    returns_dict[_t] = s.get("returns")

cols = ["count", "mean", "std", "min", "max", "sharpe", "sortino", "avg_mdd_abs"]
tbl = pd.DataFrame(all_stats).T.reindex(columns=cols)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")
print("\nFinal Summary Table (Monthly):")
print(tbl)

# Optional; quick look at aggregated equity; comment out if not needed
for k, v in returns_dict.items():
    if v is not None:
        plt.figure(figsize=(8, 5))
        plt.plot(v.cumsum(), label=k)
        plt.title(f"Cumulative Return; {k}")
        plt.legend()
        plt.grid(alpha=0.3)
        plt.tight_layout()
        plt.show()
```
